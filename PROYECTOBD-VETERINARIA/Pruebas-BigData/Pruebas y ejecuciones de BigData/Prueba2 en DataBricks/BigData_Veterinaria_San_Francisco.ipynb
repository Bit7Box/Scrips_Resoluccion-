```python
# definimos los nombres de los archivos csv de la veterinaria
archivos = [
    "CLIENTE",
    "MASCOTA",
    "VETERINARIO",
    "CITA",
    "HISTORIA_CLINICA",
    "TRATAMIENTO",
    "VACUNA",
    "PAGO"
]
dataframes = {}  # creamos un diccionario para guardar los dataframes
for archivo in archivos:  # recorremos los archivos y los cargamos con spark
    # leemos cada archivo utilizando punto y coma como separador
    df = (spark.read
        .format("csv")
        .option("header", "true")
        .option("sep", ";")
        .option("inferSchema", "true")
        .load(ruta + archivo + ".csv")
    )
    dataframes[archivo] = df    # guardamos el dataframe con el nombre de su tabla
    # mostramos la cantidad de registros y columnas
    print(archivo,"| registros:", df.count(),"| columnas:", len(df.columns))
# asignamos cada dataframe a una variable
df_cliente = dataframes["CLIENTE"]
df_mascota = dataframes["MASCOTA"]
df_veterinario = dataframes["VETERINARIO"]
df_cita = dataframes["CITA"]
df_historia = dataframes["HISTORIA_CLINICA"]
df_tratamiento = dataframes["TRATAMIENTO"]
df_vacuna = dataframes["VACUNA"]
df_pago = dataframes["PAGO"]
df_pago.printSchema()  # mostramos la estructura del dataframe de pagos
```

```python
# importamos las funciones necesarias para analizar los datos
from pyspark.sql import functions as F
# recorremos los ocho dataframes para revisar su calidad
for nombre, df in dataframes.items():
    # contamos la cantidad total de registros
    total = df.count()
    # contamos los valores nulos de cada columna
    condiciones_nulos = [
        F.sum(F.when(F.col(c).isNull(), 1).otherwise(0)).alias(c)
        for c in df.columns
    ]
    # obtenemos el resultado del conteo de nulos
    resultado_nulos = df.select(condiciones_nulos)
    # contamos los registros duplicados completos
    duplicados = total - df.dropDuplicates().count()
    # mostramos el resumen de calidad de cada tabla
    print("tabla:", nombre)
    print("total de registros:", total)
    print("registros duplicados:", duplicados)
    display(resultado_nulos)    # mostramos los valores nulos por columna
```

```python
# importamos las funciones necesarias para revisar los datos
from pyspark.sql import functions as F
# definimos las claves principales de cada tabla
claves = {
    "CLIENTE": "id_cliente",
    "MASCOTA": "id_mascota",
    "VETERINARIO": "id_veterinario",
    "CITA": "id_cita",
    "HISTORIA_CLINICA": "id_historia",
    "TRATAMIENTO": "id_tratamiento",
    "VACUNA": "id_vacuna",
    "PAGO": "id_pago"
}
# recorremos los dataframes para revisar valores vacios y claves duplicadas
for nombre, df in dataframes.items():
    # seleccionamos solamente las columnas de tipo texto
    columnas_texto = [
        campo.name
        for campo in df.schema.fields
        if campo.dataType.typeName() == "string"
    ]
    # contamos los valores vacios o que solo contienen espacios
    for columna in columnas_texto:
        vacios = df.filter(F.col(columna).isNotNull() & (F.trim(F.col(columna)) == "")).count()
        # mostramos las columnas que contienen valores vacios
        if vacios > 0:
            print(nombre, "| columna:", columna, "| vacios:", vacios)
    # identificamos la clave principal de la tabla
    clave = claves[nombre]
    # agrupamos por la clave para detectar identificadores repetidos
    duplicados_clave = (df.groupBy(clave).count().filter(F.col("count") > 1).count()
    )
    print(nombre, "| claves duplicadas:", duplicados_clave)    # mostramos el resultado de la validacion
print("revision de valores vacios y claves finalizada")
```

```python
import time  # libreria para medir los tiempos
from pyspark.sql import functions as F  # importamos las funciones de spark
# creamos una lista para almacenar los tiempos
tiempos = []
# repetimos la agregacion cinco veces
for repeticion in range(1, 6):
    inicio = time.perf_counter()    # registramos el tiempo inicial
    # agrupamos las citas por estado y contamos los registros
    resultado = (
        df_cita
        .groupBy("estado")
        .agg(F.count("*").alias("total_citas"))
        .collect()
    )
    tiempo = time.perf_counter() - inicio    # calculamos el tiempo transcurrido
    tiempos.append(tiempo)    # guardamos el tiempo obtenido
    # mostramos el resultado de cada repeticion
    print("repeticion:", repeticion,
          "| tiempo:", round(tiempo, 4), "segundos")
# calculamos el tiempo promedio de las cinco repeticiones
promedio = sum(tiempos) / len(tiempos)
# mostramos el promedio final
print("tiempo promedio:", round(promedio, 4), "segundos")
```

Prueba de escalabilidad

```python
import time  #  librerias necesarias
from pyspark.sql import functions as F
tamanos = [42, 420, 4200, 42000, 420000] # definimos el volumen de datos que utilizaremos en la prueba
# creamos una lista para guardar los resultados
resultados_escalabilidad = []
# recorremos cada tamaño de prueba
for tamano in tamanos:
    # calculamos cuantas veces debemos repetir las citas originales
    repeticiones = tamano // 42
    # generamos un dataframe sintetico sin modificar los datos originales
    df_ampliado = (
        df_cita
        .crossJoin(spark.range(repeticiones).select(F.col("id").alias("repeticion")))
        .drop("repeticion")
    )
    # creamos una lista para guardar cinco mediciones
    tiempos_prueba = []
    # repetimos la misma agregacion cinco veces
    for intento in range(5):
        # registramos el tiempo inicial
        inicio = time.perf_counter()
        # agrupamos los registros por estado
        resultado = (
            df_ampliado
            .groupBy("estado")
            .agg(F.count("*").alias("total_citas"))
            .collect()
        )
        # calculamos el tiempo de ejecucion
        tiempo = time.perf_counter() - inicio
        # guardamos el tiempo obtenido
        tiempos_prueba.append(tiempo)
    # calculamos el tiempo promedio de las cinco mediciones
    promedio = sum(tiempos_prueba) / len(tiempos_prueba)
    # guardamos los resultados del experimento
    resultados_escalabilidad.append((tamano, promedio))
    # mostramos el resultado de cada volumen
    print("registros:", tamano,
          "| tiempo promedio:", round(promedio, 4), "segundos")
# convertimos los resultados en un dataframe de spark
df_escalabilidad = spark.createDataFrame(
    resultados_escalabilidad,
    ["registros", "tiempo_promedio"]
)
display(df_escalabilidad) # tabla final para analizar los el rendimiento
```

```python
from pyspark.sql import functions as F # importamos las funciones necesarias
# filtramos solamente las citas que fueron atendidas
df_citas_atendidas = df_cita.filter(F.col("estado") == "Atendida"
)
# relacionamos las citas atendidas con los veterinarios
df_atenciones_veterinario = (
    df_citas_atendidas
    .join(
        df_veterinario,
        on="id_veterinario",
        how="inner"
    )
    .groupBy(
        "id_veterinario",
        "nombre",
        "apellido",
        "especialidad"
    )
    .agg(F.count("*").alias("total_atenciones")
    )
    .orderBy(F.desc("total_atenciones"))
)
display(df_atenciones_veterinario) # mostramos los resultados de las atenciones por veterinario
```

```python
from pyspark.sql import functions as F # funciones necesarias de spark
# agrupamos las mascotas segun su especie
df_mascotas_especie = (
    df_mascota
    .groupBy("especie")
    .agg(
        F.count("*").alias("total_mascotas")
    )
    .orderBy(F.desc("total_mascotas"))
)
# mostramos los resultados de la agrupacion
display(df_mascotas_especie)
# calculamos el total de mascotas registradas
total_mascotas = df_mascota.count()
# mostramos la cantidad total de mascotas
print("total de mascotas:", total_mascotas)
```

```python
from pyspark.sql import functions as F # funciones necesarias
# agrupamos las citas por año y mes
df_citas_mensuales = (
    df_cita
    .withColumn("mes", F.date_format("fecha_hora", "yyyy-MM"))
    .groupBy("mes")
    .agg(
        F.count("*").alias("total_citas"),
        F.sum(
            F.when(F.col("estado") == "Atendida", 1).otherwise(0)
        ).alias("citas_atendidas")
    )
    .orderBy("mes")
)
print("evolucion mensual de citas") # mostramos la evolucion mensual de citas
display(df_citas_mensuales)
# agrupamos los pagos realizados por año y mes
df_ingresos_mensuales = (
    df_pago
    .filter(F.col("estado") == "Pagado")
    .withColumn("mes", F.date_format("fecha_pago", "yyyy-MM"))
    .groupBy("mes")
    .agg(
        F.round(F.sum("monto"), 2).alias("ingresos_totales"),
        F.count("*").alias("pagos_realizados")
    )
    .orderBy("mes")
)
print("ingresos mensuales") # mostramos los ingresos mensuales de la veterinaria
display(df_ingresos_mensuales)
```

Prueba BD-02: integridad referencial

```python
# definimos las relaciones que vamos a comprobar
relaciones = [
    ("MASCOTA", "CLIENTE", "id_cliente"),
    ("CITA", "MASCOTA", "id_mascota"),
    ("CITA", "VETERINARIO", "id_veterinario"),
    ("HISTORIA_CLINICA", "CITA", "id_cita"),
    ("TRATAMIENTO", "HISTORIA_CLINICA", "id_historia"),
    ("VACUNA", "MASCOTA", "id_mascota"),
    ("VACUNA", "VETERINARIO", "id_veterinario"),
    ("PAGO", "CITA", "id_cita")
]
resultados_integridad = [] # creamos una lista para guardar los resultados
# revisamos cada relacion entre tablas
for tabla_origen, tabla_destino, clave in relaciones:
    # obtenemos los dataframes que vamos a comparar
    df_origen = dataframes[tabla_origen]
    df_destino = dataframes[tabla_destino]
    # identificamos los registros que no tienen correspondencia
    df_huerfanos = df_origen.join(df_destino.select(clave).distinct(), on=clave, how="left_anti")
    # contamos los registros originales y los huerfanos
    total = df_origen.count()
    huerfanos = df_huerfanos.count()
    # guardamos el resultado de cada relacion
    resultados_integridad.append((
        tabla_origen,
        tabla_destino,
        total,
        total - huerfanos,
        huerfanos
    ))
# creamos un dataframe con los resultados de la prueba
df_integridad = spark.createDataFrame(
    resultados_integridad,
    [
        "tabla_origen",
        "tabla_destino",
        "total_registros",
        "registros_validos",
        "registros_huerfanos"
    ]
)
display(df_integridad) # mostramos el resumen de integridad referencial
```

Prueba BD-03: limpieza con Spark DataFrames

```python
from pyspark.sql import functions as F # funciones necesarias de spark
# creamos un diccionario para guardar los dataframes limpios
dataframes_limpios = {}
# creamos una lista para guardar los resultados de limpieza
resultados_limpieza = []
# recorremos los ocho dataframes originales
for nombre, df in dataframes.items():
    # guardamos la cantidad de registros antes de limpiar
    registros_antes = df.count()
    # creamos una copia logica del dataframe original
    df_limpio = df
    # identificamos las columnas que contienen texto
    columnas_texto = [
        campo.name
        for campo in df.schema.fields
        if campo.dataType.typeName() == "string"
    ]
    # contamos las celdas de texto con espacios al inicio o al final
    celdas_corregibles = 0
    for columna in columnas_texto:
        # contamos los valores que necesitan limpieza
        cantidad = df.filter(
            F.col(columna).isNotNull() &
            (F.col(columna) != F.trim(F.col(columna)))
        ).count()
        celdas_corregibles += cantidad
        # eliminamos los espacios externos sin cambiar los valores nulos
        df_limpio = df_limpio.withColumn(columna,F.trim(F.col(columna)))
    # contamos los registros despues de la transformacion
    registros_despues = df_limpio.count()
    # guardamos el dataframe transformado
    dataframes_limpios[nombre] = df_limpio
    # registramos los resultados de la prueba
    resultados_limpieza.append((
        nombre,
        registros_antes,
        registros_despues,
        celdas_corregibles
    ))
# convertimos los resultados en un dataframe de spark
df_resumen_limpieza = spark.createDataFrame(
    resultados_limpieza,
    [
        "tabla",
        "registros_antes",
        "registros_despues",
        "celdas_con_espacios"
    ]
)
display(df_resumen_limpieza) # mostramos la comparacion antes y despues de la limpieza
```

```python
# importamos las librerias para medir tiempos y calcular estadisticas
import time
import statistics
from pyspark.sql import functions as F
# definimos los cinco volumenes de prueba
tamanos = [42, 420, 4200, 42000, 420000]
# creamos listas para guardar los resultados
mediciones = []
resumen = []
# recorremos cada volumen de datos
for tamano in tamanos:
    # generamos identificadores unicos y estados sinteticos
    df_prueba = (
        spark.range(tamano)
        .select(
            F.col("id").alias("id_cita"),
            F.when(
                F.col("id") % 42 == 41,
                "Pendiente"
            ).otherwise("Atendida").alias("estado")
        )
        .cache()
    )
    # materializamos los registros antes de medir
    df_prueba.count()
    # guardamos los tiempos de las cinco ejecuciones
    tiempos = []
    for repeticion in range(1, 6):
        # registramos el inicio de la agregacion
        inicio = time.perf_counter()
        # agrupamos los registros por estado
        resultado = (
            df_prueba
            .groupBy("estado")
            .agg(F.count("*").alias("total_citas"))
            .collect()
        )
        # calculamos el tiempo transcurrido
        tiempo = time.perf_counter() - inicio
        # guardamos la medicion individual
        tiempos.append(tiempo)
        mediciones.append((tamano, repeticion, tiempo))
        print(
            "registros:", tamano,
            "| repeticion:", repeticion,
            "| tiempo:", round(tiempo, 4)
        )
    # calculamos las estadisticas de cada volumen
    resumen.append((
        tamano,
        statistics.mean(tiempos),
        min(tiempos),
        max(tiempos),
        statistics.stdev(tiempos)
    ))
    # liberamos los datos almacenados en cache
    df_prueba.unpersist()
# creamos el dataframe de mediciones individuales
df_mediciones_final = spark.createDataFrame(
    mediciones,
    ["registros", "repeticion", "tiempo_segundos"]
)
# creamos el dataframe con las estadisticas
df_resumen_final = spark.createDataFrame(
    resumen,
    [
        "registros",
        "promedio",
        "minimo",
        "maximo",
        "desviacion_estandar"
    ]
)
# mostramos los resultados para el informe
display(df_resumen_final)
```
